```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

plt.rcParams['font.family'] = ['Malgun Gothic', 'DejaVu Sans']
plt.rcParams['axes.unicode_minus'] = False

%matplotlib inline

import warnings
warnings.filterwarnings('ignore')
```

```python
train_df = pd.read_csv('C:/Users/LG/OneDrive/Desktop/Jupyter/space_titanic_data/space_titanic_data_file/train.csv')
test_df = pd.read_csv('C:/Users/LG/OneDrive/Desktop/Jupyter/space_titanic_data/space_titanic_data_file/test.csv')

print("Train Data Shape: ", train_df.shape)
print("Test Data Shape: ", test_df.shape)

print("Train Data", display(train_df.head(5)))
print("Test Data", display(test_df.head(5)))

train_df.info()

display(train_df.describe())
```

## 변수 설명

PassengerId - 각 승객의 고유 ID입니다. gggg_pp 형식을 가짐.
(gggg: 함께 여행하는 그룹의 번호입니다. (가족인 경우가 많습니다)
pp: 그룹 내에서의 번호입니다.)

HomePlanet - 승객이 출발한 행성(일반적으로 거주지)임 (예: Earth, Europa, Mars)

CryoSleep - 승객이 항해 기간 동안 냉동 수면 상태를 선택했는지 여부. 냉동 수면 중인 승객은 객실 내에 머물게 됨

Cabin - 승객이 머무는 객실 번호로, deck/num/side 형식을 가짐
(deck: 데크 위치 (A, B, C, D, E, F, G, T)
num: 객실 번호)

side - 우주선의 측면 (P: Port/좌현, S: Starboard/우현)

Destination - 승객이 내릴 목적지 행성

Age - 승객의 나이

VIP - 승객이 항해 중 특별 VIP 서비스를 위해 추가 요금을 지불했는지 여부

RoomService - 룸서비스 이용액

FoodCourt - 푸드코트 이용액

ShoppingMall - 쇼핑몰 이용액

Spa - 스파 이용액

VRDeck - VR 데크 이용액

Name: 승객의 성과 이름

Transported (Target) - 승객이 다른 차원으로 이동되었는지 여부. 이 데이터셋에서 예측해야 하는 타겟 변수

## 변수 시각화

```python
#데이터 타입 변경
mapping = {True:1, False:0}

train_df['Transported'] = train_df['Transported'].map(mapping)

train_df['CryoSleep'] = train_df['CryoSleep'].fillna(False).astype(int)
test_df['CryoSleep'] = test_df['CryoSleep'].fillna(False).astype(int)

train_df['VIP'] = train_df['VIP'].fillna(False).astype(int)
test_df['VIP'] = test_df['VIP'].fillna(False).astype(int)
```

```python
print('Transported Value Counts:')
print(train_df['Transported'].value_counts())

print('\nDestination Value Counts:')
print(train_df['Destination'].value_counts())

plt.figure() # Transported 시각화
sns.countplot(x='Transported', data=train_df)
plt.title('Transported Dstribution (1 = True, 0 = False)')
plt.show()

plt.figure() # CryoSleep 에 따른 Transported 값 시각화
sns.countplot(x='CryoSleep', hue='Transported', data=train_df)
plt.title('Transported by CryoSleep')
plt.show()

plt.figure() # Destination 에 따른  Transported 값 시각화
sns.countplot(x='Destination', hue='Transported', data=train_df)
plt.title('Transported by Destination')
plt.show()

plt.figure() # Destination에 따른 CryoSleep 값 시각
sns.countplot(x='Destination', hue='CryoSleep', data=train_df)
plt.title('Destination by CryoSleep')
plt.show()
```

## 전처리

```python
# 결측치 처리
train_df = train_df.dropna(axis=0)
test_df = test_df.dropna(axis=0)
```

```python
# Cabin deck, num, side 쪼개기
train_df[['Cabin_deck', 'Cabin_num', 'Cabin_side']] = train_df['Cabin'].str.split('/', expand=True)
train_df.drop(columns=['Cabin'], inplace=True)

test_df[['Cabin_deck', 'Cabin_num', 'Cabin_side']] = test_df['Cabin'].str.split('/', expand=True)
test_df.drop(columns=['Cabin'], inplace=True)

# Cabin_num이 문자열로 인식되니 int로 바꿔줌.
train_df['Cabin_num'] = train_df['Cabin_num'].replace('None', -1).astype(int)
test_df['Cabin_num'] = train_df['Cabin_num'].replace('None', -1).astype(int)
```

```python
# 비행선 서비스 비용이니 TotalPrice라는 새로운 변수를 만들어서 보는 것이 현명할 수 있다고 생각함.
train_df['TotalPrice'] = train_df['RoomService'] + train_df['FoodCourt'] + train_df['ShoppingMall'] + train_df['Spa'] + train_df['VRDeck']
test_df['TotalPrice'] = test_df['RoomService'] + test_df['FoodCourt'] + test_df['ShoppingMall'] + test_df['Spa'] + test_df['VRDeck']

# Name 컬럼은 핵심변수에 영향을 줄 수 없음.
drop_col = ['Name', 'RoomService', 'FoodCourt', 'ShoppingMall', 'VRDeck', 'Spa', 'VIP']
train_df.drop(columns=drop_col, inplace=True)
test_df.drop(columns=drop_col, inplace=True)
```

```python
# Destination 레이블 인코딩-로우값이 문자열이면 인식X -> 0, 1, 2 로 레이블을 인코딩하기 위함.
from sklearn.preprocessing import LabelEncoder
le = LabelEncoder()

train_df['Destination'] = le.fit_transform(train_df['Destination'])
test_df['Destination'] = le.transform(test_df['Destination'])
# 레이블이 0,1,2 로 되어서 선형회귀 모델은 0<1<2 가중치로 이해할 수 있음. 하지만 Xgboost 또는 다른 트리 모델에 있어서 좋은 효과를 보임.
```

```python
# 원-핫 인코딩을 사용. 
Categorical_columns = ['HomePlanet', 'Cabin_deck', 'Cabin_side']

train_df = pd.get_dummies(train_df, columns=Categorical_columns, prefix=Categorical_columns)
test_df = pd.get_dummies(test_df, columns=Categorical_columns, prefix=Categorical_columns)

prefixes = tuple(Categorical_columns)

# 혹시 train과 test의 더미 컬럼 불일치에 대비해 동일한 컬럼 맞춰주기
for col in train_df.columns:
    if col.startswith(prefixes) and col not in test_df.columns: 
        test_df[col] = 0 # test에 없는 train 더미변수 생성

for col in test_df.columns:
    if col.startswith(prefixes) and col not in train_df.columns:
        train_df[col] = 0 # train 에 없는 test 더미변수 생성 

# 모든 피처 컬럼을 동일한 순서로 정렬(Transported 제외)
feature_cols = [col for col in train_df.columns if col != 'Transported']
train_df = train_df[['Transported'] + feature_cols]
test_df = test_df[feature_cols]
```

```python
print('Train Columns: ', train_df.columns.tolist())
print('Test Columns: ', test_df.columns.tolist())
print('Train Shape: ', train_df.shape, 'Test Shape: ', test_df.shape)
```

## 추가 전처리

Cabin_side_P, Cabin_side_S 와 Transported의 상관관계 절댓값이 둘 다 0.11 임. 따라서 둘 중 하나를 삭제하여도 모델학습에 지장이 없을 것.
오히려 과적합이될 가능성이 적어질듯.

선형적인 데이터과 비선형적인 데이터를 만들어서 회귀모델과 분류모델을 각각 만들어보도록 함.

선형적 데이터 -> 다중공선성 제거, 수치형 피쳐 스케일링(로그변환 등)
비선형적 데이터 -> 변수를 합쳐서 새로운 변수 만들기, 

### 비선형 데이터 만들기

```python
train_tree = train_df.copy()
test_tree = test_df.copy()

# 새로운 파생 변수 생성
# 냉동 수면중인 사람은 소비가 0원, 하지만 냉동수면이 아닌 승객중에 지출이 0원인 사람들도 존재 -> 지출여부 변수 생성
train_tree['NoSpend'] = (train_tree['TotalPrice'] == 0).astype(int)
test_tree['NoSpend'] = (test_tree['TotalPrice'] == 0).astype(int)

# 나이(Age) 구간화 - pd.cut(data, bis, labels) 사용하여 데이터 구간화
train_tree['AgeGroup'] = pd.cut(train_tree['Age'], bins=[-1, 12, 19, 35, 60, 100], labels=[0, 1, 2, 3, 4])
test_tree['AgeGroup'] = pd.cut(test_tree['Age'], bins=[-1, 12, 19, 35, 60, 100], labels=[0, 1, 2, 3, 4])

train_tree['AgeGroup'] = train_tree['AgeGroup'].fillna(2).astype(int) # 가장 많은 연령대인 2번(청년)으로 채움
test_tree['AgeGroup'] = test_tree['AgeGroup'].fillna(2).astype(int)

# Cain_num 구간화 - ID 와 깊은 연관 -> 객실 번호가 비슷한 사람들은 같은 구역에 모여있었을 가능성이 있음.
train_tree['Cabin_num'] = train_tree['Cabin_num'].fillna(0)
test_tree['Cabin_num'] = test_tree['Cabin_num'].fillna(0)

train_tree['CabinZone'] = (train_tree['Cabin_num'] // 150).astype(int)
test_tree['CabinZone'] = (test_tree['Cabin_num'] // 150).astype(int)

# 상관관계가 아주 낮은 컬럼과 처리한 컬럼 삭제
drop_cols_tree = ['PassengerId', 'Cabin_deck_T', 'Cabin_side_P', 'Age', 'Cabin_num']
train_tree.drop(columns=drop_cols_tree, inplace=True)
test_tree.drop(columns=drop_cols_tree, inplace=True)
```

### 선형 데이터 만들기

```python
# 다중공선성 확인용 VIF(분산 팽창 지수)
# 히트맵은 두 변수간 관계만 보여주지만, VIF는 하나의 변수가 나머지 모든 변수들의 조합으로 설명되는가?"를 측정(수치형 변수)

from statsmodels.stats.outliers_influence import variance_inflation_factor

train_linear = train_df.copy()
test_linear = test_df.copy()

# 수치형 데이터 추출
vif_data = train_linear.select_dtypes(include=[np.number])

# VIF 계산
vif_df = pd.DataFrame()
vif_df["feature"] = vif_data.columns
vif_df["VIF"] = [variance_inflation_factor(vif_data.values, i) for i in range(len(vif_data.columns))]

# VIF 시각화
print(vif_df.sort_values(ascending=False, by="VIF"))
```

```python
# TotalPrice 로그변환 - log1p (log(1+x)) 를 사용하여 로그변환함
# 대부분 0이고 최댓값과 차이가 너무 많기에 로그변환을 통해 편차를 맞춰줌
train_linear['TotalPrice'] = np.log1p(train_linear['TotalPrice'])
test_linear['TotalPrice'] = np.log1p(train_linear['TotalPrice'])

# 수치형 피처 스케일링 - StandardScaler 를 사용하여 모든 수치형 변수의 평균을 0, 표준편차를 1에 맞춤
# Age 와 TotalPrice 단위의 차가 너무 많이 나고 TotalPrice가 단위가 더 커서 모델이 TotalPrice에 더 초점을 맞추기 때문
from sklearn.preprocessing import StandardScaler

scaler = StandardScaler()
num_features =['Age', 'TotalPrice']

train_linear[num_features] = scaler.fit_transform(train_linear[num_features])
test_linear[num_features] = scaler.transform(test_linear[num_features])

# 불필요한 컬럼 제거
drop_cols_linear = ['PassengerId', 'Cabin_side_P', 'Cabin_deck_T','Cabin_deck_A', 'Cabin_num'] # Cabin_num 은 비선형적
train_linear.drop(columns=drop_cols_linear, inplace=True)
test_linear.drop(columns=drop_cols_linear, inplace=True)
```
